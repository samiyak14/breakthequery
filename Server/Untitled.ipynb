```python
import sqlite3 as sq3
connection = sq3.connect('database.db')
cursor = connection.cursor()
cursor.execute("DELETE FROM QuizData;")
cursor.execute("SELECT * FROM QuizData;")
print(cursor.fetchall())
cursor.execute("SELECT Name, COUNT(QuestionNumber) as QuestionsSolved, SUM(TimeTaken) FROM QuizData GROUP BY Name ORDER BY QuestionsSolved DESC;")
results = cursor.fetchall()
for result in results:
    for r in result:
        print(f"{r}\t\t", end="")
    print()
```
